## Quickstart: draw water
Try a simple cartesian coordinates to test both 2D and 3D rendering.

```python
import py3Dmol
import sys

sys.path.append("..")

import src.io.read_file as rf
from src.molecule.molecule import Molecule

name_input="../resources/input.vel"

# Read the config
config = rf.config_man(name_input)

#Obtain the name of the molecules
keys = list(config.keys())

print(keys)
```

```python
# Example: Water molecule (H2O) in XYZ format
molec="water" #chouce a obtion of the previous list
mol = Molecule(molec).from_hocon(config, molec)
xyz = repr(mol)

# Create a viewer
view = py3Dmol.view(width=600, height=500)

# Load the XYZ coordinates
view.addModel(xyz, 'xyz')

# Choose a style (stick, sphere, or line)
view.setStyle({'stick': {}})

# Auto-zoom to fit molecule
view.zoomTo()

# Show interactive 3D
view.show()
```
